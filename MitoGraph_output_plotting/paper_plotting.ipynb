```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the the data

```python
df = pd.read_csv('230915_all_conditions_decon/output-summary.csv')

# Split the dataframe for the individual plots
TRAK_df = df[df['Condition'].isin(['si control', 'TRAK1 kd', 'TRAK2 kd', 'wt', 'TRAK1 oe', 'TRAK2 oe'])]
gal_df = df[df['Condition'].isin(['wt', 'galactose O/N', 'galactose 48h'])]
```

List of parameters you can plot

```python
for item in df.columns:
    print(item)
```

### All TRAK plot

```python
# Choose the parameter you would like to plot
parameter = 'MitoGraph_Connectivity_Score'

# Color scheme for the plot
colors = ['#82c6e2', '#6acc64', '#d65f5f', '#4878d0', '#6acc64', '#d65f5f']

plt.figure(figsize=(4, 4))
sns.boxplot(TRAK_df, x='Condition', y=parameter, showfliers=False, palette=(sns.color_palette(colors)), linewidth=1)
sns.swarmplot(TRAK_df, x='Condition', y=parameter, legend=False, color='darkslategray', size=5, alpha=0.5, marker='o')
plt.xlabel('')
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.ylabel('MitoGraph Connectivity Score', fontsize=14)
plt.yticks(fontsize=12)
plt.show()
#plt.savefig('paper_plots/TRAK_morphology.png', dpi=300, bbox_inches='tight')
```

### All Galactose plot

```python
colors = ['#4878d0', '#956cb4', '#dc7ec0']

plt.figure(figsize=(2, 4))
sns.boxplot(gal_df, x='Condition', y=parameter, showfliers=False, palette=(sns.color_palette(colors)), linewidth=1, order=['wt', 'galactose O/N', 'galactose 48h'])
sns.swarmplot(gal_df, x='Condition', y=parameter, legend=False, color='darkslategray', size=5, alpha=0.5, marker='o', order=['wt', 'galactose O/N', 'galactose 48h'])
plt.xlabel('')
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.ylabel('MitoGraph Connectivity Score', fontsize=14)
plt.yticks(fontsize=12)

plt.show()
#plt.savefig('paper_plots/galactose_morphology.png', dpi=300, bbox_inches='tight')
```
