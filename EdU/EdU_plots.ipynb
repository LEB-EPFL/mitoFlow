```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('EdU_averages_plot.csv')
pixel_size = 0.11
```

Calculate the correct distance by multiplying with the microscopes pixel size

```python
df['Dist_corr'] = df['Dist']*pixel_size
```

```python
df.head()
```

```python
custom_palette = ['#0BD566', '#F27B40']

fig, ax = plt.subplots(figsize=(4,4))
sns.boxplot(data=df, x='Metadata_Time', y='Dist_corr', hue='EdUNormSignal', showfliers=False, palette=custom_palette)
sns.swarmplot(data=df, x='Metadata_Time', y='Dist_corr', hue='EdUNormSignal', dodge=True, color='slategray', linewidth=1, alpha=0.7, size=4, legend=False)
plt.ylim(df.Dist_corr.min()-1, 22)
plt.ylabel('Distance to MTOC (µm)', fontsize=14)
plt.yticks([8,12,16,20], fontsize=12)
plt.xticks(fontsize=12)
plt.xlabel('time (min)', fontsize=14)
plt.legend(title='')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
#plt.savefig('EdU_distance_plot.png', dpi=300, bbox_inches='tight')
```
