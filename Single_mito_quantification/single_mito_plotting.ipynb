```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the data

```python
data = pd.read_csv('Relative_distance_mitos.csv')
```

```python
data.head()
```

```python
l_palette = ('royalblue', 'lightsteelblue')
```

```python
fig, ax = plt.subplots(figsize=(5,4))

sns.lineplot(data=data, x='Time', y='Yellow_Green_RelDistance', hue='Type', palette=l_palette, linewidth=1.5, alpha=0.8, legend=False)
sns.scatterplot(data=data, x='Time', y='Yellow_Green_RelDistance', hue='Type', palette=l_palette, size='Type', sizes=(10,10), alpha=0.5)

plt.ylabel('red/green median distance (A.U.)', fontsize=14)
plt.xlabel('time (min)', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title='')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
#plt.savefig('single_mito_median_distance.png', dpi=300, bbox_inches='tight')
```

# Load the data for the mito area plot

```python
mito_area_data = pd.read_csv('Mean_area_plot.csv')
pixel_size = 0.103
```

```python
mito_area_data['Area_microns'] = mito_area_data['Area'] * pixel_size**2
```

```python
c_palette = ('#7cba3f', '#df7bb1')

fig, ax = plt.subplots(figsize=(6.5,4))

sns.boxplot(mito_area_data, y='Area_microns', x='Time', hue='Colour', palette=c_palette , showfliers=False)
#sns.swarmplot(mito_area_data, y='Area', x='Time', hue='Colour', color='black', dodge=True, size=4, alpha=0.5)
sns.stripplot(mito_area_data, y='Area_microns', x='Time', hue='Colour', palette=('#276419', '#940457'), alpha=0.4, dodge=True, size=3, jitter=0.2, legend=False)
plt.xlabel('time (min)', fontsize=14)
plt.ylabel('mitochondrial area (μm²)', fontsize=14)
plt.ylim(0,4)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend().remove()

ax.tick_params('x', length=0, width=0, which='major')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)

plt.show()
#plt.savefig('single_mito_area.png', dpi=300, bbox_inches='tight')
```
